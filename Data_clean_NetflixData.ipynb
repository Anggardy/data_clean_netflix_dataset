## Data Null

```python
df.dropna(subset=['director', 'cast'], inplace=True)
```

```python
df['country'] = df['country'].fillna('Unknown Country')
df['rating'] = df['rating'].fillna('Unrated')
df['duration'] = df['duration'].fillna('Not yet known')
```

## Formatting

```python
df['date_added'] = df['date_added'].str.strip()
df['date_added'] = pd.to_datetime(df['date_added'])
```

# Partition Data

```python
df_movie = df[df['type'] == 'Movie']
df_series = df[df['type'] == 'TV Show']
```

```python
df_movie.head()
```

```python
df_series.head()
```

```python
df_series['season_count'] = df_series['duration'].str.split(' ').str[0]
```

```python
df_movie['duration_value'] = df_movie['duration'].str.split(' ').str[0]
```

# Data to CSV

```python
output_path = '/content/netflix_titles_processed.csv'
df.to_csv(output_path, index=False)
print(f"DataFrame saved to {output_path}")
```
